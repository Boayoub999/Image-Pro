# Thresholding

```python
import cv2

# Load the image in grayscale
img = cv2.imread('Parrot.jpg', cv2.IMREAD_GRAYSCALE)

# Set the threshold values
threshold_values = [0, 50, 100, 150, 200]

cv2.imshow('Original Image', img)

for threshold_value in threshold_values:
    ret, thresh = cv2.threshold(img, threshold_value, 255, cv2.THRESH_BINARY)
    cv2.imshow(f'Thresholded Image ({threshold_value})', thresh)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Histogram Processing

```python
import cv2
import matplotlib
import matplotlib.pyplot as plt
import numpy as np
from skimage import data, img_as_float
from skimage import exposure
```

```python
img = cv2.imread('Parrot.jpg', cv2.IMREAD_GRAYSCALE)
p2, p98 = np.percentile(img, (2, 98))
img_rescale = exposure.rescale_intensity(img, in_range=(p2, p98))
```

```python
fig = plt.figure(figsize=(10, 7)) 
fig.add_subplot(2, 1, 1) 
plt.imshow(img, cmap = 'gray') 
plt.axis('off') 
plt.title('Original image') 
fig.add_subplot(2, 1, 2) 
plt.hist(img.flat, bins = 256, range=(0, 255)) 
plt.xlabel('gray level values') 
plt.ylabel('Number of pixels') 
plt.show() 



fig = plt.figure(figsize=(10, 7)) 	
fig.add_subplot(2, 1, 1) 
plt.imshow(img_rescale, cmap = 'gray') 
plt.axis('off') 
plt.title('Contrast stretch')
fig.add_subplot(2, 1, 2) 
plt.hist(img_rescale.flat, bins = 256, range=(0, 255)) 
plt.xlabel('gray level values') 
plt.ylabel('Number of pixels') 
plt.show()
```

# Task#1

```python
import matplotlib.pyplot as plt
import numpy as np
from skimage import data, exposure

img = data.moon()

p3, p80 = np.percentile(img, (3, 80))

img_rescale = exposure.rescale_intensity(img, in_range=(p3, p80))

fig = plt.figure(figsize=(10, 7))

fig.add_subplot(2, 1, 1)
plt.imshow(img_rescale, cmap='gray')
plt.axis('off')
plt.title('Contrast Stretch (3rd & 80th Percentiles)')

fig.add_subplot(2, 1, 2)
plt.hist(img_rescale.flat, bins=256, range=(0, 255))
plt.xlabel('Gray Level Values')
plt.ylabel('Number of Pixels')

plt.tight_layout()
plt.show()
```

# Task#2

```python
img = data.moon()

img_eq = exposure.equalize_hist(img)

fig = plt.figure(figsize=(10, 7))

fig.add_subplot(2, 1, 1)
plt.imshow(img_eq, cmap='gray')
plt.axis('off')
plt.title('Equalized Image (Flattened Histogram)')

fig.add_subplot(2, 1, 2)
plt.hist(img_eq.flat, bins=256)
plt.xlabel('Pixel Intensity')
plt.ylabel('Number of Pixels')

plt.tight_layout()
plt.show()
```

# Task#3

```python
from skimage.exposure import match_histograms

reference = data.rocket()
image = data.chelsea()

matched = match_histograms(image, reference, channel_axis=-1)

fig, (ax1, ax2, ax3) = plt.subplots(nrows=1, ncols=3, figsize=(15, 5))

ax1.imshow(image)
ax1.set_title('Source Image (Chelsea)')
ax1.axis('off')

ax2.imshow(reference)
ax2.set_title('Reference Image (Rocket)')
ax2.axis('off')

ax3.imshow(matched)
ax3.set_title('Matched Image')
ax3.axis('off')

plt.tight_layout()
plt.show()
```
